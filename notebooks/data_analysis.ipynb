# Bharat Electoral Analytics — Exploratory Data Analysis (EDA)

**Title:** Election Results and Voter Behaviour: An Interactive D3.js Analysis of Indian Elections (1952–2024)  
**Objective:** Quantitative statistical analysis of voter turnout trajectories, gender convergence, regional alliance patterns, and victory margin competitiveness across 18 Lok Sabha elections.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

PROC_DIR = '../data/processed'
sns.set_theme(style="whitegrid", font="sans-serif")
plt.rcParams['font.size'] = 11
```

## 1. Load Processed Datasets

```python
df_nat = pd.read_csv(os.path.join(PROC_DIR, 'national_series.csv'))
df_party = pd.read_csv(os.path.join(PROC_DIR, 'party_performance.csv'))
df_reg = pd.read_csv(os.path.join(PROC_DIR, 'regional_shifts.csv'))
df_state = pd.read_csv(os.path.join(PROC_DIR, 'state_demographics.csv'))

print("Datasets loaded successfully.")
```

## 2. Longitudinal Voter Turnout & Gender Crossover Analysis (1951–2024)

Key observation: In 1971, the gender gap was -11.79 pp (male turnout 60.90% vs female 49.11%). By 2019, female turnout (67.18%) crossed male turnout (67.01%), maintaining close parity in 2024 (65.78% female vs 65.55% male).

```python
plt.figure(figsize=(11, 5.5))
plt.plot(df_nat['year_num'], df_nat['turnout_pct'], marker='o', color='#0f172a', lw=2.5, label='Overall Turnout')
plt.plot(df_nat['year_num'], df_nat['male_turnout_pct'], marker='s', color='#2563eb', lw=1.8, label='Male Turnout')
plt.plot(df_nat['year_num'], df_nat['female_turnout_pct'], marker='^', color='#e11d48', lw=2, label='Female Turnout')

plt.title('Lok Sabha Historical Voter Turnout & Gender Trajectory (1952–2024)', fontsize=13, weight='bold', pad=12)
plt.xlabel('Election Year')
plt.ylabel('Turnout Percentage (%)')
plt.ylim(40, 75)
plt.legend(frameon=True)
plt.tight_layout()
plt.show()
```

## 3. Party Performance & Alliance Seat Conversion

Analysis of 2024 Lok Sabha: BJP won 240 seats (36.56% vote share); INC won 99 seats (21.19% vote share).

```python
top_parties = df_party.head(10).copy()
plt.figure(figsize=(10, 5))
colors = {'NDA': '#ea580c', 'INDIA': '#2563eb', 'Unaligned': '#0d9488', 'Mixed': '#64748b'}
bar_colors = [colors.get(a, '#94a3b8') for a in top_parties['alliance']]

plt.barh(top_parties['party'], top_parties['seats_won'], color=bar_colors)
plt.axvline(272, color='#dc2626', linestyle='--', label='Majority Mark (272 Seats)')
plt.gca().invert_yaxis()
plt.title('Top 10 Parties by Seats Won (2024 Lok Sabha)', fontsize=12, weight='bold')
plt.xlabel('Seats Won')
plt.legend()
plt.tight_layout()
plt.show()
```

## 4. Regional Alliance Coalitions & Voter Participation

```python
reg_df = df_reg.set_index('region')[['nda_seats', 'india_seats', 'other_seats']]
ax = reg_df.plot(kind='bar', figsize=(11, 5), color=['#ea580c', '#2563eb', '#0d9488'], width=0.75)
plt.title('Regional Alliance Seat Distribution (2024 Lok Sabha)', fontsize=12, weight='bold')
plt.ylabel('Seats Won')
plt.xticks(rotation=0)
plt.legend(['NDA', 'INDIA', 'Others'])
plt.tight_layout()
plt.show()
```

## 5. State-Level Demographic Insights & Victory Margins

```python
print("Highest Turnout States/UTs:")
print(df_state[['state_ut', 'seats', 'turnout_pct', 'youth_elector_share_pct', 'avg_victory_margin_pct']].sort_values(by='turnout_pct', ascending=False).head(5))

print("\nMost Competitive States/UTs (Lowest Victory Margin):")
print(df_state[['state_ut', 'seats', 'turnout_pct', 'avg_victory_margin_pct']].sort_values(by='avg_victory_margin_pct').head(5))
```
